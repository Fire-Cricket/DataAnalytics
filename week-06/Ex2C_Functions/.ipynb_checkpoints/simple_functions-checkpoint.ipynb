```python
def favorite_things(name, movies, music):
    print(f'My name is {name}. '
          f'My favorite movie is {movies}. '
          f'My favorite song is {music}. '
)
    
```

```python
print(favorite_things('Jordan Worrobah', 'X-Men Apocalypse', 'Crooked Smiles'))
```

```python
def favorite_place(place, why):
    return f'One of my favorite places is {place} because {why}'
```

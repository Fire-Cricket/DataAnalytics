```python
def get_soc_sec_tax(gross_pay):
    social_security_tax = gross_pay * 0.062
    return social_security_tax
```

```python
def get_medicare_tax(gross_pay):
    medicare_tax = gross_pay * 0.0145
    return medicare_tax
```

```python
def get_federal_tax(gross_pay, withholding_code):

    match withholding_code:

        case 0:
            tax_rate = 0.23

        case 1:
            tax_rate = 0.21

        case 2:
            tax_rate = 0.195

        case 3:
            tax_rate = 0.185

        case 4:
            tax_rate = 0.18

        case _:
            over_four = withholding_code - 4

            tax_rate = 0.18 - (over_four * 0.005)

    federal_tax = gross_pay * tax_rate

    return federal_tax
                
```

# Person1: gross pay $750 withholding code 0

```python
print(get_soc_sec_tax(750))
```

```python
print(get_medicare_tax(750))
```

```python
print(get_federal_tax(750, 0))
```

```python
print(get_federal_tax(1550, 2))
```

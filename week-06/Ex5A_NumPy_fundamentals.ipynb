## Array Basics

```python
fruits = ['apple', 'banana', 'cherry', 'date']
```

```python
arr1 = np.array(fruits)
```

```python
print(arr1)
```

```python
print(arr1.dtype)
```

```python
# The dtype returned is <U6.
# The U means Unicode string.
# The 6 means the longest string in the array has 6 characters.
```

```python
arr2 = np.array([(10, 11, 12), (13, 14, 15)])
```

```python
print(arr2)
```

```python
print(arr2.T)
```

```python
arr3 = np.array([(1, 3, 5), (2, 4, 6)])
```

```python
print(arr3)
```

```python
arr4 = np.array([arr3[0, 2], arr3[1, 1]])
```

```python
print(arr4)
```

```python
arr5 = np.ones((2, 2, 2))
```

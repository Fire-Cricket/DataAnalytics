## Generating Random Numbers

```python
r_1 = np.random.randint(1, 21, size=2)
print(r_1)
```

```python
r_2 = np.random.randint(1, 21, size=(1,2))
print(r_2)
```

```python
r_3 = np.random.rand(2,1)
print(r_3)
```

```python
r_4 = np.random.randint(20, 101, size=(4,5))
print(r_4)
```

```python
r_5 = np.random.rand(4,5)
print(r_5)
```

## Generating Random Choices from

```python
students = np.array(('Leon Poulson', 'Ivana Nwaobi', 'Kenneth Jeffreys',
                      'Shawn Sandman', 'Sincere Jackson', 'Cris Ramiez', 
                      'Lily Wells', 'Dani Trawick', 'Jeremiah Butts', 
                      'Devin Hollingsworth', 'Ebony Dupree', 'Sameul Veloz'))
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)
```

This wouldn't be an effective way to create breakout rooms because it does not protect duplicates. It is possible to have the same person in two diffrent rooms, while it is also possible to have one person not in any rooms.

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4,5))
print(bell_curve)
```

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop goes through the first dimension of the array, which are the rows. 
Each row is multiplied by 2 as a group, so the output stays in array form.

The second loop goes through the second dimension, which are the individual elements inside each row. 
Each number is multiplied and printed separately instead of as part of an array.

```python
for x in arrA:
    for y in x:
        loop_result = y * 2
        print(loop_result)
```

This loop accesses each individual element in the second dimension,
so each number is doubled and printed separately instead of as rows.

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))

print(stacked_array_2)
```

This gives an error because the arrays do not have the same shape.
joined_array_1 has shape (4,3) and joined_array_2 has shape (2,6).
np.stack() requires arrays to have identical dimensions.

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)
print(another_2)
```

This error happens because the arrays have different numbers of rows.
To concatenate on axis=1, both arrays must have the same size along axis=0.

```python
hstack = np.hstack((arrA, arrA))
print(hstack)
```

Stacking Along Rows

```python
vstack = np.vstack((arrA, arrA))
print(vstack)
```

Stacking Along Columns

```python
dstack = np.dstack((arrA, arrA))
print(dstack)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)
```

The result contains 2 arrays.
Each array is 2-dimensional and contains 2 rows and 3 columns.

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
```

The result contains 3 arrays.
Each array is still 2-dimensional, but the arrays do not all have the same number of rows.
NumPy splits the rows as evenly as possible.

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6], [13, 5], [7, 3]])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

```python
y = arrB >= 7
arrD = arrB[y]
print(arrD)
```

### Insights from info()

The dataset contains 262 records and 5 text-based columns: title, description, category, language, and actor_name. Since all columns are strings, the most useful information from describe() is the count, unique, top, and frequency values. These statistics help identify whether there are duplicate values, the most common categories, languages, or actors, and how diverse the data is.

Some summarizations are less useful. The title and description fields are likely intended to be unique for most records, so knowing the most common title or description does not provide much analytical value. Since all columns are text-based, there are no numerical statistics such as mean, median, minimum, or maximum available.

The file appears to be unnormalized because actor names, categories, and languages are stored directly within the same table as the movie information. In a normalized database design, these attributes would likely be separated into related tables and connected through keys to reduce redundancy and improve data integrity.

# Exercise 1.C Performing operations on arrays

```python
import numpy as np
```

## Sample Arrays

```python
r_1 = np.random.randint(1, 21, size=2)

r_2 = np.random.randint(1, 21, size=(1,2))

r_3 = np.random.rand(2,1)

r_4 = np.random.randint(20, 101, size=(4,5))

r_5 = np.random.rand(4,5)


print("r_1:\n", r_1)
print()
print("nr_2:\n", r_2)
print()
print("nr_3:\n", r_3)
print()
print("nr_4:\n", r_4)
print()
print("nr_5:\n", r_5)
```

## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

It looks like the numbers are just shuffling up in down in their columns. It does not look like the shuffle is applying to shuffling between columns only between rows.

```python
new_shuffle = np.random.permutation(r_4)
print('\n new_shuffle:')
print(new_shuffle)
print('\n r_4:')
print(r_4)
```

You would use shuffle when you want to randomly rearrange the original array itself. 
You would use permutation when you want a shuffled copy while keeping the original array unchanged.
